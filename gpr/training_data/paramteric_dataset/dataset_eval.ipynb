```python
import os
import sys

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

```

```python
filename = "simulated_dataset.csv"

df = pd.read_csv(filename)

```

```python
plt.figure(figsize=(10, 6))
columns_to_plot = ['x','t','L_fusion', 'die_temp_l', 'die_temp_r', 'temp']
fig, axs = plt.subplots(2, 3, figsize=(14, 10))

for ax, col in zip(axs.flatten(), columns_to_plot):
    ax.plot(df.index, df[col])
    ax.set_title(col)
    ax.set_xlabel('Index')
    ax.set_ylabel(col)

plt.tight_layout()
plt.show()
```
